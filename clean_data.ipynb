```python
import pandas as pd

# Find the path of the csv file
import os
print(os.getcwd())
```

```python
# import coral.csv
global_bleaching_environmental_csv = pd.read_csv("/Users/zhanghanjing/Desktop/ENGG212/Data/global_bleaching_environmental.csv")
print(global_bleaching_environmental_csv.head())
```

```python
# check the columns
print(list(global_bleaching_environmental_csv.columns))
```

```python
# find the lose value(columns)
nd_counts = global_bleaching_environmental_csv.astype(str).apply(

    lambda col: col.str.strip().str.lower().eq("nd").sum()

)
pd.set_option("display.max_rows", None)
print(nd_counts[nd_counts > 0].sort_values(ascending=False))
```

```python
# delete raw
cleaned_raw = global_bleaching_environmental_csv.drop(columns=[

    "Site_Comments",
    "Sample_Comments",
    "Bleaching_Comments",
    "Reef_ID",
    "Substrate_Name",
    "Percent_Cover",
    "Bleaching_Level"
])

cleaned_raw.to_csv("output_clean.csv", index=False)
```

```python
# recheck the columns
print(list(cleaned_raw.columns))
```

```python
# find the lose value(columns) again
nd_counts_in_cleaned_raw = cleaned_raw.astype(str).apply(

    lambda col: col.str.strip().str.lower().eq("nd").sum()

)
pd.set_option("display.max_rows", None)
print(nd_counts_in_cleaned_raw[nd_counts_in_cleaned_raw > 0].sort_values(ascending=False))
```

```python
# 删除任意列中值等于 nd 的整行
mask_nd = cleaned_raw.astype(str).apply(
    lambda col: col.str.strip().str.lower().eq("nd")
).any(axis=1)

cleaned_raw_no_nd = cleaned_raw[~mask_nd]

print("Number of rows before deletion:", len(cleaned_raw))
print("Number of Rows Containing “nd”:", mask_nd.sum())
print("Number of rows after deletion", len(cleaned_raw_no_nd))
```

```python
nd_counts = cleaned_raw_no_nd.astype(str).apply(

    lambda col: col.str.strip().str.lower().eq("nd").sum()

)

print("剩余 nd 总数:", nd_counts.sum())
```

```python
# check the columns title
print(cleaned_raw_no_nd.head())
```

```python
features = [

    "Distance_to_Shore",
    "Turbidity",
    "Cyclone_Frequency",
    "Depth_m",
    "Temperature_Kelvin",
    "Windspeed",
    "SSTA",
    "SSTA_Maximum",
    "SSTA_Frequency",
    "SSTA_DHW"

]


cleaned_raw_no_nd = cleaned_raw_no_nd[features]


print(cleaned_raw_no_nd.head())
```

```python
# save the csv file
cleaned_raw_no_nd.to_csv("coral_cleaned.csv", index=False)
```
